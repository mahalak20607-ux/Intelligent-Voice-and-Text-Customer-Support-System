```python
# ============================================================
# VOICEDESK AI
# Intelligent Voice & Text Customer Support System
# ============================================================

import gradio as gr
import speech_recognition as sr
import tempfile
import os
import re
import random
from datetime import datetime


# ============================================================
# TICKET STORAGE
# ============================================================

tickets = []


# ============================================================
# AUDIO CONVERSION
# ============================================================

def convert_to_wav(audio_file):

    audio = AudioSegment.from_file(audio_file)

    audio = audio.set_channels(1)
    audio = audio.set_frame_rate(16000)
    audio = audio.set_sample_width(2)

    wav_file = tempfile.NamedTemporaryFile(
        suffix=".wav",
        delete=False
    ).name

    audio.export(
        wav_file,
        format="wav"
    )

    return wav_file


# ============================================================
# SPEECH TO TEXT
# ============================================================

def speech_to_text(audio_file):

    if audio_file is None:
        return ""

    wav_file = None

    try:

        wav_file = convert_to_wav(
            audio_file
        )

        recognizer = sr.Recognizer()

        with sr.AudioFile(wav_file) as source:

            audio_data = recognizer.record(
                source
            )

        text = recognizer.recognize_google(
            audio_data,
            language="en-US"
        )

        return text

    except sr.UnknownValueError:

        return "Could not understand the voice."

    except sr.RequestError:

        return "Speech recognition service unavailable."

    except Exception as e:

        return f"Speech error: {str(e)}"

    finally:

        if (
            wav_file
            and os.path.exists(wav_file)
        ):

            os.remove(wav_file)


# ============================================================
# CATEGORY DETECTION
# ============================================================

def detect_category(text):

    text = text.lower()

    categories = {

        "Internet / Network": [
            "internet",
            "wifi",
            "network",
            "connection",
            "broadband",
            "signal"
        ],

        "Payment / Billing": [
            "payment",
            "bill",
            "billing",
            "refund",
            "money",
            "charge",
            "transaction"
        ],

        "Account / Login": [
            "login",
            "password",
            "account",
            "sign in",
            "otp",
            "verification"
        ],

        "Product / Service": [
            "product",
            "service",
            "order",
            "delivery",
            "broken",
            "damaged"
        ],

        "Technical Support": [
            "error",
            "bug",
            "crash",
            "not working",
            "problem",
            "issue",
            "failure"
        ]
    }

    for category, keywords in categories.items():

        for keyword in keywords:

            if keyword in text:

                return category

    return "General Support"


# ============================================================
# PRIORITY DETECTION
# ============================================================

def detect_priority(text):

    text = text.lower()

    high_words = [
        "urgent",
        "emergency",
        "immediately",
        "critical",
        "blocked",
        "cannot access",
        "not working",
        "fraud",
        "stolen"
    ]

    medium_words = [
        "slow",
        "problem",
        "issue",
        "failed",
        "delay",
        "error"
    ]

    for word in high_words:

        if word in text:

            return "🔴 High"

    for word in medium_words:

        if word in text:

            return "🟠 Medium"

    return "🟢 Low"


# ============================================================
# SENTIMENT
# ============================================================

def analyze_sentiment(text):

    try:

        result = sentiment_model(
            text[:512]
        )[0]

        label = result["label"]
        score = result["score"]

        if label == "POSITIVE":

            return f"😊 Positive ({score:.0%})"

        return f"😟 Negative ({score:.0%})"

    except:

        return "Neutral"


# ============================================================
# SUGGESTED TEAM
# ============================================================

def suggested_team(category):

    teams = {

        "Internet / Network":
            "Network Support Team",

        "Payment / Billing":
            "Billing & Finance Team",

        "Account / Login":
            "Account Support Team",

        "Product / Service":
            "Service Operations Team",

        "Technical Support":
            "Technical Support Team",

        "General Support":
            "Customer Care Team"
    }

    return teams.get(
        category,
        "Customer Care Team"
    )


# ============================================================
# CREATE SUPPORT TICKET
# ============================================================

def create_ticket(text):

    if not text or not text.strip():

        return (
            "⚠️ Please enter a complaint.",
            "",
            "",
            "",
            ""
        )

    text = text.strip()

    category = detect_category(
        text
    )

    priority = detect_priority(
        text
    )

    sentiment = analyze_sentiment(
        text
    )

    team = suggested_team(
        category
    )

    ticket_id = (
        "VD-"
        + str(
            random.randint(
                1000,
                9999
            )
        )
    )

    created_time = datetime.now().strftime(
        "%d-%m-%Y %I:%M %p"
    )

    ticket = {

        "Ticket ID": ticket_id,
        "Issue": text,
        "Category": category,
        "Priority": priority,
        "Sentiment": sentiment,
        "Team": team,
        "Status": "Open",
        "Created": created_time
    }

    tickets.append(
        ticket
    )

    result = f"""
### 🎫 Support Ticket Created

**Ticket ID:** `{ticket_id}`

**Category:** {category}

**Priority:** {priority}

**Sentiment:** {sentiment}

**Assigned Team:** {team}

**Status:** 🟢 Open

**Created:** {created_time}
"""

    return (
        result,
        ticket_id,
        category,
        priority,
        team
    )


# ============================================================
# VOICE PROCESSING
# ============================================================

def process_voice(audio):

    if audio is None:

        return (
            "⚠️ Please record your complaint.",
            ""
        )

    text = speech_to_text(
        audio
    )

    if not text:

        return (
            "⚠️ No speech detected.",
            ""
        )

    if text.startswith(
        "Could not"
    ) or text.startswith(
        "Speech error"
    ):

        return (
            f"❌ {text}",
            ""
        )

    return (
        "### 📝 Voice Complaint Transcription\n\n"
        + text,
        text
    )


# ============================================================
# DASHBOARD
# ============================================================

def dashboard():

    if not tickets:

        return """
### 📊 Support Dashboard

No support tickets created yet.
"""

    total = len(tickets)

    open_count = sum(
        1
        for t in tickets
        if t["Status"] == "Open"
    )

    high_count = sum(
        1
        for t in tickets
        if "High" in t["Priority"]
    )

    rows = ""

    for ticket in tickets:

        rows += f"""
| {ticket["Ticket ID"]} | {ticket["Category"]} | {ticket["Priority"]} | {ticket["Status"]} |
"""

    return f"""
### 📊 VoiceDesk AI Dashboard

**Total Tickets:** {total}

**Open Tickets:** {open_count}

**High Priority:** {high_count}

---

| Ticket ID | Category | Priority | Status |
|---|---|---|---|
{rows}
"""


# ============================================================
# PROFESSIONAL CSS
# ============================================================

custom_css = """

body {
    background:
        linear-gradient(
            135deg,
            #eef2ff,
            #f5f3ff
        );
}

.gradio-container {
    max-width: 1200px !important;
    margin: auto;
}

.title {
    text-align: center;
    font-size: 42px;
    font-weight: 800;
    color: #4f46e5;
}

.subtitle {
    text-align: center;
    font-size: 18px;
    color: #555;
    margin-bottom: 25px;
}

button {
    border-radius: 10px !important;
}

textarea,
input {
    border-radius: 12px !important;
}

"""


# ============================================================
# APPLICATION
# ============================================================

with gr.Blocks(
    css=custom_css,
    title="VoiceDesk AI"
) as app:

    # --------------------------------------------------------
    # HEADER
    # --------------------------------------------------------

    gr.Markdown(
        """
        <div class="title">
        🎧 VoiceDesk AI
        </div>

        <div class="subtitle">
        Intelligent Voice & Text Customer Support
        Management System
        </div>
        """
    )


    # ========================================================
    # TEXT SUPPORT
    # ========================================================

    with gr.Tab(
        "📝 Text Support"
    ):

        text_input = gr.Textbox(
            label="Customer Complaint",
            placeholder=(
                "Describe your problem here..."
            ),
            lines=7
        )

        create_button = gr.Button(
            "🎫 Create Support Ticket",
            variant="primary"
        )

        ticket_result = gr.Markdown()

        with gr.Row():

            ticket_id_output = gr.Textbox(
                label="Ticket ID"
            )

            category_output = gr.Textbox(
                label="Category"
            )

        with gr.Row():

            priority_output = gr.Textbox(
                label="Priority"
            )

            team_output = gr.Textbox(
                label="Assigned Team"
            )

        create_button.click(
            create_ticket,
            inputs=text_input,
            outputs=[
                ticket_result,
                ticket_id_output,
                category_output,
                priority_output,
                team_output
            ]
        )


    # ========================================================
    # VOICE SUPPORT
    # ========================================================

    with gr.Tab(
        "🎤 Voice Support"
    ):

        gr.Markdown(
            """
            ### 🎙️ Submit a Voice Complaint

            Record your complaint and VoiceDesk AI
            will process it into a support request.
            """
        )

        voice_input = gr.Audio(
            sources=[
                "microphone",
                "upload"
            ],
            type="filepath",
            label="🎙️ Record Customer Complaint"
        )

        voice_button = gr.Button(
            "🔊 Process Voice",
            variant="primary"
        )

        voice_transcription = gr.Markdown()

        voice_text = gr.Textbox(
            label="Processed Complaint",
            lines=5
        )

        voice_button.click(
            process_voice,
            inputs=voice_input,
            outputs=[
                voice_transcription,
                voice_text
            ]
        )


        voice_create_button = gr.Button(
            "🎫 Create Ticket From Voice",
            variant="primary"
        )

        voice_result = gr.Markdown()

        voice_ticket_id = gr.Textbox(
            label="Generated Ticket ID"
        )

        voice_category = gr.Textbox(
            label="Category"
        )

        voice_priority = gr.Textbox(
            label="Priority"
        )

        voice_team = gr.Textbox(
            label="Assigned Team"
        )

        voice_create_button.click(
            create_ticket,
            inputs=voice_text,
            outputs=[
                voice_result,
                voice_ticket_id,
                voice_category,
                voice_priority,
                voice_team
            ]
        )


    # ========================================================
    # DASHBOARD
    # ========================================================

    with gr.Tab(
        "📊 Support Dashboard"
    ):

        dashboard_button = gr.Button(
            "🔄 Refresh Dashboard",
            variant="primary"
        )

        dashboard_output = gr.Markdown()

        dashboard_button.click(
            dashboard,
            outputs=dashboard_output
        )


    # ========================================================
    # FOOTER
    # ========================================================

    gr.Markdown(
        """
        ---

        ### 🚀 VoiceDesk AI

        **Voice Support • Text Support •
        Complaint Classification • Priority Detection •
        Ticket Management • Support Analytics**

        Intelligent customer support powered by
        AI and natural language processing.
        """
    )


# ============================================================
# LAUNCH
# ============================================================

app.launch(
    share=True,
    debug=True
)
```
